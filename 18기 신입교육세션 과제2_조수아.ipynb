# **insight 정리**
 1. Pclass,Cabin,Emarked,Fare는 모두 위치에 관련된 변수이다.
    - 가설 : Pclass가 높은 등급일 경우 생존률이 더 높을것이다
    - 가설1(Cabin-Pclass) : Cabin이 A(상층)에 위치할수록 Pclass가 더 높을 것이다(객실등급1이 더 많이 존재)

      / Pclass가 높은 객실일수록 생존률이 더 높다.
      /A,B(최상층) C,D(상층) E(중간) F,G(하층) T(최하층)
    - 가설2(Fare-Pclass) : Fare이 높은 객실일수록 Pclass가 높을 것이다

       / Fare이 높을수록 생존률이 더 높다.
    - 가설3(Embarked-Pclass) : Embarked별 Pclass 분포가 다르면, Pclass가 높은 승객이 많이 탑승한 Embarked일수록 생존률이 높을 것이다

       / Pclass가 1등급인 티켓을 가진 사람이 많이 탑승한 Embarked일수록 생존률이 더 높을 것이다.
 2. family(SipSp+Parch)
    - 가설 : 동반 가족수가 많은수록 생존률이 낮을 것이다

# **2. Age 변수 결측값 채우기**

```python
data1.columns
```

```python
data1['Name'] ## 결측치를 Name으로 유추해서 하기 
data1
```

- Mr: 18세 이상 남성
- Miss : 미혼 여성
- Mrs : 기혼 여성
- Master : 13세이하 남성
- Dr : 박사, 의사 

```python
data1_age_yes = data1[~data1['Age'].isnull()]
print(data1_age_yes) #Age가 있는 행 
```

```python
data1_age_no = data1[data1['Age'].isnull()]
print(data1_age_no) #Age가 NaN인 행 
```

```python
data1['Title'] = data1['Name'].str.split(",").str[1].str.split(".").str[0]
print(data1['Title'])
data1['Title'].unique()
```

```python
data1["Age_NA"] = data1["Age"].isnull()
data1.groupby("Title")["Age_NA"].sum().sort_values() 
#결측치가 있는 변수가 Dr, Master, Mrs, Miss, Mr라는것을 확인 
```

```python
#Mr 결측치 채우기 
mask = data1['Title'] == ' Mr'
data1.loc[mask, 'Age'] = (data1.loc[mask].groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.mean())))
data1.groupby("Title")['Age'].apply(lambda x: x.isnull().sum()) #결측치 사라진거 확인 
```

```python
#Miss 결측치 채우기 
mask = data1['Title'] == ' Miss'
data1.loc[mask, 'Age'] = (data1.loc[mask].groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.mean())))
data1.groupby("Title")['Age'].apply(lambda x: x.isnull().sum()) #결측치 사라진거 확인 
```

```python
data1[(data1['Title'] == ' Mrs') & (data1['Age'].isnull())]
```

```python
#Mrs 결측치 채우기 
mask = data1['Title'] == ' Mrs'
data1.loc[mask, 'Age'] = (data1.loc[mask].groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.mean())))
```

```python
#Master 결측치 채우기 
mask = data1['Title'] == ' Master'
data1.loc[mask, 'Age'] = (data1.loc[mask].groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.mean())))
```

```python
#Dr 결측치 채우기 
mask = data1['Title'] == ' Dr'
data1.loc[mask, 'Age'] = (data1.loc[mask].groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.mean())))
```
